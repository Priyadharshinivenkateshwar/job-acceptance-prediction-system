```python
df = pd.read_csv("../data/HR_Job_Placement_Dataset.csv")
df_clean = df.copy()
```

```python
text_cols = ['gender', 'internship_experience', 'company_tier', 'degree_specialization']
for col in text_cols:
  df_clean[col] = df_clean[col].astype(str).str.strip().str.title()
```

```python
cat_fills = {
    'career_switch_willingness': 'Not Willing',
    'relevant_experience': 'Not Relevant', 
    'job_role_match': 'Not Matched',
    'layoff_history': 'No',
    'relocation_willingness': 'Not Willing'
}
for col, fill_value in cat_fills.items():
    df_clean[col] = df_clean[col].fillna(fill_value)
```

```python
perc_cols = ['ssc_percentage', 'hsc_percentage', 'degree_percentage', 'skills_match_percentage']
score_cols = ['technical_score', 'aptitude_score', 'communication_score']
```

```python
for col in perc_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
```

```python
for col in score_cols:
    df_clean[col] = df_clean[col].fillna(0)
```

```python
exp_cols = ['certifications_count', 'years_of_experience', 'employment_gap_months']
for col in exp_cols:
    df_clean[col] = df_clean[col].fillna(0)
```

```python
for col in perc_cols + score_cols:
    df_clean[col] = df_clean[col].clip(0, 100)
```

```python
df_clean['age_years'] = df_clean['age_years'].clip(18, 65)
```

```python
initial_rows = len(df_clean)
df_clean = df_clean.drop_duplicates()
print(f"Removed {initial_rows - len(df_clean)} duplicates")
```

```python
df_clean['status'] = df_clean['status'].fillna('Not Placed')  
df_clean['status_binary'] = (df_clean['status'] == 'Placed').astype(int)
```

```python
df_clean.to_csv("../data/cleaned_data.csv", index=False)
```
